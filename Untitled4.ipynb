```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Data
cus_names = ["Akhi", "Asri", "anu", "ammu", "smiley"]
internet_speed = [2, 10, 1, 15, 3]
payment_failure = [3, 0, 4, 0, 2]
delivery_charges = [200, 50, 250, 40, 180]
purchase_completed = [False, True, False, True, False]

# Create DataFrame
df = pd.DataFrame({
    "Customer": cus_names,
    "Internet_Speed": internet_speed,
    "Payment_Failure": payment_failure,
    "Delivery_Charges": delivery_charges,
    "Purchase_Completed": purchase_completed
})

print("Customer Dataset")
print(df)

# Total abandonment
total_abandonment = purchase_completed.count(False)
print("\nTotal Abandonment:", total_abandonment)

# ---------------- INTERNET DIAGNOSIS ----------------
print("\nInternet Diagnosis")
for i in range(len(cus_names)):
    if internet_speed[i] < 5:
        print(cus_names[i], internet_speed[i], "Mbps")

# ---------------- PAYMENT DIAGNOSIS ----------------
print("\nPayment Diagnosis")
for i in range(len(cus_names)):
    if payment_failure[i] > 0:
        print(cus_names[i], payment_failure[i])

# ---------------- DELIVERY DIAGNOSIS ----------------
print("\nDelivery Diagnosis")
for i in range(len(cus_names)):
    if delivery_charges[i] > 200:
        print(cus_names[i], delivery_charges[i])

# ---------------- ROOT CAUSE ANALYSIS ----------------
print("\nRoot Cause Analysis")
for i in range(len(cus_names)):
    if purchase_completed[i] == False:
        if internet_speed[i] < 5:
            print(cus_names[i], "abandonment cause due to low internet speed")
        elif payment_failure[i] > 0:
            print(cus_names[i], "abandonment cause due to payment failure")
        elif delivery_charges[i] > 150:
            print(cus_names[i], "abandonment cause due to high delivery charges")

# ---------------- VISUALIZATION ----------------
df.plot(x="Customer", kind="bar", figsize=(10,5))

plt.title("Customer Dataset Analysis")
plt.ylabel("Values")
plt.xticks(rotation=0)
plt.show()
```

```python
import pandas as pd
# Load dataset
file_path = "/content/internet.csv"
df = pd.read_csv(file_path)

# Remove unwanted column
if 'Unnamed: 0' in df.columns:
    df = df.drop('Unnamed: 0', axis=1)

# Convert ping column to numeric
# Errors become NaN

df['ping'] = pd.to_numeric(df['ping'], errors='coerce')

# Remove rows with missing values

df = df.dropna()

# ---------- SIGNAL QUALITY ----------
# Lower ping = Better signal

def signal_quality(ping):
    if ping <= 50:
        return 'Good'
    elif ping <= 100:
        return 'Average'
    else:
        return 'Bad'

# Create signal column

df['Signal'] = df['ping'].apply(signal_quality)

# ---------- CUSTOMER CHURN ----------
# Simple logic:
# Low speed or bad signal -> customer may change company
# Otherwise customer may continue

def churn(speed, signal):
    if speed < 1000 or signal == 'Bad':
        return 'Will Change Company'
    else:
        return 'Continue Same Company'

# Create churn column

df['Customer_Status'] = df.apply(
    lambda x: churn(x['kb/s'], x['Signal']),
    axis=1
)

# ---------- TOTAL COUNTS ----------
status_count = df['Customer_Status'].value_counts()

print("\nCustomer Status Count:\n")
print(status_count)

# ---------- COMPANY PERFORMANCE ----------
company_speed = df.groupby('operator')['kb/s'].mean()

print("\nAverage Internet Speed by Company:\n")
print(company_speed)

# ---------- GRAPH 1 ----------
# Customer churn graph

status_count.plot(kind='bar')
plt.title('Customer Churn Analysis')
plt.xlabel('Customer Status')
plt.ylabel('Number of Customers')
plt.show()

# ---------- GRAPH 2 ----------
# Average speed graph

company_speed.plot(kind='bar')
plt.title('Average Internet Speed by Company')
plt.xlabel('Company')
plt.ylabel('Average Speed (kb/s)')
plt.show()
```
